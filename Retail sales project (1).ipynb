### 2.1 distribution of sales amount

```python
plt.figure(figsize=(8,5))
sns.histplot(df["Sales_Amount"])
plt.title("distribution of sales amount")
plt.xlabel("Sales_amount")
plt.ylabel("Frequency")
plt.show()
```

#### interpretation
This histogram shows that most sales amounts are low,while only a few transactions have high sales amount.The distribution is positively skewed,indicating that high-values sales are less frequent.

### 2.2 product category distribution 

```python
plt.figure(figsize=(8,5))
sns.countplot(x="Product_Category",data=df)
plt.title("product category distribution")
plt.xlabel("product category")
plt.ylabel("count")
plt.xticks(rotation=45)
plt.show()
```

#### interpetation
The chart shows that all product categories have almost the same number of sales records.

### 2.3 customer type distribution 

```python
plt.figure(figsize=(7,5))
sns.countplot(x="Customer_Type",data=df)
plt.title("product category distribution")
plt.xlabel("customer_type")
plt.ylabel("count")
plt.show()
```

#### interpretation
The chart shows that both new and returning customers are almost equally dsitributed.

### 2.4 Payment method distribution

```python
plt.figure(figsize=(7,5))
sns.countplot(x="Payment_Method",data=df)
plt.title("payment method distribution")
plt.xlabel("payment_method")
plt.ylabel("count")
plt.xticks(rotation=45)
plt.show()
```

#### interpreation
The chart shows that card,bank transfer,cash payments are used almost equally by customers.


### 2.5 sales channel distribution 

```python
plt.figure(figsize=(6,6))
df["Sales_Channel"].value_counts().plot(kind="pie",autopct="%1.1f%%")
plt.title("sales channel distribution")
plt.ylabel("")
plt.show()
```

#### interpreation
The pie chart shows that online and offline sales are almost equally distributed.This indicates a balanced contribution from both sales channel

### 2.6 Region distribution

```python
plt.figure(figsize=(8,5))
sns.countplot(x="Region",data=df)
plt.title("Region distribution")
plt.xlabel("Region")
plt.ylabel("count")
plt.xticks(rotation=45)
plt.show()
```

#### interpretation
The chart show that sales records are almost equally distributed across all regions.

## 3.3 Region vs sales amount 

```python
plt.figure(figsize=(8,5))
sns.barplot(x="Region",y="Sales_Amount",data=df)
plt.title("Region vs sales amount")
plt.xlabel("Region")
plt.ylabel("Average Sales Amount")
plt.xticks(rotation=65)
plt.show()
```

### intrepretation

## 3.4 quantity sold vs sales amount

```python
plt.figure(figsize=(7,5))
sns.scatterplot(x="Quantity_Sold",y="Sales_Amount",data=df)
plt.title("quantity vs sales amount")
plt.show()
```

### interpretation
The scatter plot shows a strong positive relationship between quantity sold and sales amount.As quantity sold increase,sales amount also increases.

## 3.5 Discount vs Sales Amount 

```python
plt.figure(figsize=(7,5))
sns.scatterplot(x="Discount",y="Sales_Amount",data=df)
plt.title("Discount vs sales amount")
plt.show()
```

### interpretation 
The scatter plot shows a negative relationship between Discount and sales amount .As the discount increases, the sales amount generally decreases.

## Multivariate Analysis

#### 4.1 Product Category vs Region Vs Sales Amount

```python
plt.figure(figsize=(8,5))
sns.barplot(x="Product_Category",y="Sales_Amount",hue="Region",data=df)
plt.title("Product Category vs Region Vs Sales Amount")
plt.xticks(rotation=45)
plt.show()
```

#### Interpretation
The chart shows that average sales are nearly the same across all product categories and regions,with only slight differences.

### 4.2 customer Type vs Payment Method vs Sales Amount

```python
plt.figure(figsize=(8,5))
sns.barplot(x="Customer_Type",y="Sales_Amount",hue="Payment_Method",data=df)
plt.title("customer Type vs Payment Method vs Sales Amount")
plt.show()
```

### interpretation
The chart shows that new and returning customers have similar average sales across all payment methods.Difference are very small.

### 4.3 Correlation Heatmap

```python
plt.figure(figsize=(6,4))
sns.heatmap(df[["Sales_Amount","Quantity_Sold","Discount"]].corr(),annot=True,cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

### Interpretation
Quantity Sold has a moderate positive relationship with sales amount(0.65),while Discount has a very weak negative relationship(-0.12).

#### Conclusion
This Exploratory Data Analysis (EDA) of the retail sales dataset provided valuable insights into sales performance and customer purchasing patterns. The dataset was found to be clean and suitable for analysis, with no missing values or duplicate records. The analysis showed that sales vary across different product categories and regions, indicating that these factors have a significant impact on business performance. It was also observed that higher discounts are generally associated with lower sales amounts, suggesting a negative relationship between discount and sales. Overall, the findings demonstrate that EDA is an effective approach for identifying trends, understanding sales behaviour, and supporting data-driven decision-making. These insights can help retailers improve sales strategies, optimize pricing, and enhance overall business performance.
